```python
import time
from numba import njit

def profiling(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Runtime of {func.__name__}: {end - start:.6f} seconds")
        return result
    return wrapper

@profiling
def normal_function(simulations):
    total = 0
    for i in range(simulations):
        total += i ** 2
    return total

@njit
def numba_function(simulations):
    total = 0
    for i in range(simulations):
        total += i ** 2
    return total

simulations = 1000000

result1 = normal_function(simulations)
print("Normal function result:", result1)

start = time.time()
result2 = numba_function(simulations)
end = time.time()

print("Numba function result:", result2)
print(f"Runtime of Numba function: {end - start:.6f} seconds")
```
