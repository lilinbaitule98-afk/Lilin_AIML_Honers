```python
import csv

class Employee:

    def add_employee(self):
        empid = input("Enter Employee ID: ")
        name = input("Enter Employee Name: ")
        salary = input("Enter Salary: ")

        with open("employee.csv", "a", newline="") as f:
            writer = csv.writer(f)
            writer.writerow([empid, name, salary])

        print("Employee added successfully")

    def display_employee(self):
        try:
            with open("employee.csv", "r") as f:
                reader = csv.reader(f)

                for row in reader:
                    print(row)

        except FileNotFoundError:
            print("File not found")

    def update_employee(self):
        empid = input("Enter Employee ID to update: ")

        with open("employee.csv", "r") as f:
            rows = list(csv.reader(f))

        found = False

        for row in rows:
            if row[0] == empid:
                row[1] = input("Enter new Employee Name: ")
                row[2] = input("Enter new Salary: ")
                found = True

        with open("employee.csv", "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerows(rows)

        if found:
            print("Employee updated successfully")
        else:
            print("Employee not found")

    def delete_employee(self):
        empid = input("Enter Employee ID to delete: ")

        with open("employee.csv", "r") as f:
            rows = list(csv.reader(f))

        new_rows = []
        found = False

        for row in rows:
            if row[0] == empid:
                found = True
            else:
                new_rows.append(row)

        with open("employee.csv", "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerows(new_rows)

        if found:
            print("Employee deleted successfully")
        else:
            print("Employee not found")


obj = Employee()

while True:
    print("\n1. Add Employee")
    print("2. Display Employees")
    print("3. Update Employee")
    print("4. Delete Employee")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        obj.add_employee()
    elif choice == "2":
        obj.display_employee()
    elif choice == "3":
        obj.update_employee()
    elif choice == "4":
        obj.delete_employee()
    elif choice == "5":
        break
    else:
        print("Invalid choice")
```
