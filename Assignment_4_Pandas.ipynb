```python
import pandas as pd

file = "employee.csv"

while True:

    print("\n1. Add Employee")
    print("2. Display Employees")
    print("3. Update Employee")
    print("4. Delete Employee")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":

        empid = input("Enter Employee ID: ")
        name = input("Enter Employee Name: ")
        salary = input("Enter Salary: ")

        try:
            df = pd.read_csv(file)
        except:
            df = pd.DataFrame(columns=["EmployeeID", "Name", "Salary"])

        new_employee = pd.DataFrame({
            "EmployeeID": [empid],
            "Name": [name],
            "Salary": [salary]
        })

        df = pd.concat([df, new_employee], ignore_index=True)
        df.to_csv(file, index=False)

        print("Employee added successfully")

    elif choice == "2":

        try:
            df = pd.read_csv(file)
            print(df)
        except:
            print("File not found")

    elif choice == "3":

        empid = input("Enter Employee ID to update: ")

        try:
            df = pd.read_csv(file)

            if empid in df["EmployeeID"].astype(str).values:
                name = input("Enter new Employee Name: ")
                salary = input("Enter new Salary: ")

                df.loc[df["EmployeeID"].astype(str) == empid, "Name"] = name
                df.loc[df["EmployeeID"].astype(str) == empid, "Salary"] = salary

                df.to_csv(file, index=False)

                print("Employee updated successfully")
            else:
                print("Employee not found")

        except:
            print("File not found")

    elif choice == "4":

        empid = input("Enter Employee ID to delete: ")

        try:
            df = pd.read_csv(file)

            if empid in df["EmployeeID"].astype(str).values:
                df = df[df["EmployeeID"].astype(str) != empid]
                df.to_csv(file, index=False)

                print("Employee deleted successfully")
            else:
                print("Employee not found")

        except:
            print("File not found")

    elif choice == "5":
        break

    else:
        print("Invalid choice")
```
