```python
def output_decorator(operation):
    def decorator(func):
        def wrapper(a, b):
            result = func(a, b)
            print("#" * 26)
            print(f"# {operation}")
            print(f"# input: {a}, {b}")
            print(f"# output: {result}")
            print("#" * 26)
        return wrapper
    return decorator


@output_decorator("Adding")
def add(a, b):
    return a + b


add(4, 3)
```
